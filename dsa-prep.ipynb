```python
def take_input(input_str):
    inp = [int(_) for _ in input_str.split()]

    n = inp[0]
    arr = inp[1:n+1]
    p = inp[n+1]
    q = inp[n+2]
    queries = []
    for i in range(n+3, n+3+2*q, 2):
        queries.append([inp[i], inp[i+1]])

    return n, arr, p, q, queries
```

```python
def get_ans(n, arr, p, q, queries):
    left = [0] * n
    right = [0] * n
    
    left[0] = arr[0]
    for i in range(1, n):
        left[i] = math.gcd(left[i-1], arr[i])
    
    right[n-1] = arr[n-1]
    for i in range(n-2, -1, -1):
        right[i] = math.gcd(right[i+1], arr[i])
    
    ans = 0
    for ind, val in queries:
        ind -= 1
        l = left[ind-1] if ind > 0 else right[i+1]
        r = right[i+1] if ind < n-1 else left[ind-1]
        new_gcd = math.gcd(l, val, r)
        ans += new_gcd == p
    
    return ans
```
