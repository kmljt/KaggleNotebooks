```python
import pandas as pd
```

### Reading CSV

```python
df = pd.read_csv(
        '/kaggle/input/datasets/olgagmiufana1/fragrantica-com-fragrance-dataset/fra_perfumes.csv'
        # '/kaggle/input/datasets/olgagmiufana1/fragrantica-com-fragrance-dataset/fra_cleaned.csv'
)

df
```

### Extracting Name and Brand from Description

```python
df.Description
```

```python
df[['Name', 'Brand']] = df.Description.str.extract(r'^(.*?)by(.*?)is ')

df[['Name', 'Brand']]
```

```python
df
```

### Fixing Gender

```python
df.Gender.value_counts()
```

```python
df.Gender = df.Gender.str.lower().map({
    'for men': 'M',
    'for women': 'F',
    'for women and men': 'U'
})

df.Gender
```

```python
df.Gender.value_counts()
```

```python
df
```

```python
df
```

### Converting `Rating Count` to `Int64`

```python
df['Rating Count'] = (
    df['Rating Count']
    .str.replace(',', '')
    .astype(float)
    .astype('Int64')
)
```

```python
df.dtypes
```

```python
df
```

### Filtering Columns

```python
df = df[['Brand', 'Name', 'Gender', 'Rating Value', 'Rating Count', 'Main Accords', 'url']]

df
```

```python
df.columns = ['brand', 'name', 'gender', 'rating_value', 'rating_count', 'main_accords', 'url']

df.columns
```

```python
df
```

### Handling NULLs

```python
df.isna().sum()
```

```python
df = df.dropna(subset=['brand'])

df.isna().sum()
```

```python
df[df.rating_value.isna()]
```

```python
df[['rating_value', 'rating_count']] = df[['rating_value', 'rating_count']].fillna(0)

df.isna().sum()
```

### Finally, ANALYSIS!

```python
df = df.iloc[:, :5]

df
```

```python
(
    df[(df.rating_count > 1000) & (df.gender != 'F')]
        .sort_values('rating_value', ascending=False)
        .head(30)
)
```

```python
(
    df[
        (True)
        & (df.brand == 'Lattafa Perfumes') 
        & (df.rating_count > 1000) 
        & (df.gender != 'F')
    ]
    .sort_values('rating_value', ascending=False)
    .head(30)
)
```
